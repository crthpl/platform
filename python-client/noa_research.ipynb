```python
import pandas as pd

# Load data
orders = pd.read_csv('historical_data/orders.csv', parse_dates=['timestamp'])
trades = pd.read_csv('historical_data/trades.csv', parse_dates=['transaction_timestamp'])

# Process each market
market_grids = {}
for market in orders['market_name'].unique():
    # Filter market data
    mkt_orders = orders[orders['market_name'] == market].set_index('timestamp')
    mkt_trades = trades[trades['market_name'] == market].set_index('transaction_timestamp')
    
    # Create minute grids
    order_grid = mkt_orders.groupby(pd.Grouper(freq='1Min')).last()
    trade_grid = mkt_trades.groupby(pd.Grouper(freq='1Min')).agg({
        'price': 'last',
        'size': 'last'
    }).rename(columns={'price': 'last_trade_price', 'size': 'last_trade_size'})
    
    # Combine and store
    market_grids[market] = pd.concat([order_grid, trade_grid], axis=1).fillna(method='ffill')

# Convert transaction_id to int where present
for market in market_grids:
    if 'transaction_id' in market_grids[market].columns:
        market_grids[market]['transaction_id'] = market_grids[market]['transaction_id'].astype('Int64')


# Example: Look at first market's data
first_market = list(market_grids.keys())[0]
print(f"\nSample of {first_market} market grid:")
display(market_grids[first_market].head())
```

```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd

alpha = market_grids['alpha']
diffs = alpha['last_trade_price'] - alpha['last_trade_price'].shift(1)
WINDOW_SIZE = 8

# Get top 20 largest price changes, dropping any duplicates
top_diffs = diffs.abs().sort_values(ascending=False).head(20)
top_timestamps = top_diffs.index.drop_duplicates()

print("\nTimestamps of largest price changes:")
for timestamp in top_timestamps:
    print(timestamp)

# Create a figure to plot price changes after each top timestamp
fig = go.Figure()

# For each top timestamp, plot the next WINDOW_SIZE minutes of relative price changes
for timestamp in top_timestamps:
    window_end = timestamp + pd.Timedelta(minutes=WINDOW_SIZE)
    window = alpha[timestamp:window_end]
    
    # Skip if no data in window
    if len(window) == 0:
        continue
        
    # Calculate price changes relative to the price at the top timestamp
    ref_price = alpha.loc[timestamp, 'last_trade_price']
    price_changes = window['last_trade_price'] - ref_price
    
    # Create a time-normalized x-axis (minutes from start)
    minutes_from_start = [(t - timestamp).total_seconds()/60 for t in window.index]
    
    # Determine color based on if final price is higher or lower
    final_change = price_changes.iloc[-1] if len(price_changes) > 0 else 0
    line_color = 'green' if final_change > 0 else 'red'
    
    # Plot this window, ensuring unique name for each trace
    fig.add_trace(go.Scatter(
        x=minutes_from_start,
        y=price_changes.values,
        name=f'From {timestamp.strftime("%H:%M:%S")}',
        mode='lines',
        line=dict(color=line_color),
        showlegend=True
    ))

fig.update_layout(
    title='Price Changes After Largest Moves',
    xaxis_title='Minutes After Event',
    yaxis_title='Price Change',
    showlegend=True
)

fig.update_xaxes(range=[0, WINDOW_SIZE])
fig.show()
```
